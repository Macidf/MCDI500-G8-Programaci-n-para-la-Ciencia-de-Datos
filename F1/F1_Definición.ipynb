## 1. Contexto y problemática

La acuicultura depende de mantener el agua dentro de rangos compatibles con la vida de los peces:
la temperatura regula su metabolismo, el oxígeno disuelto condiciona su supervivencia y compuestos
como el amonio son tóxicos en concentraciones altas. Los sensores IoT permiten registrar estas
variables de forma continua, pero trasladan el problema desde la medición hacia la **calidad del
registro**.

El proyecto trabaja con **Pondsdata** (Arepalli, s.f.): lecturas de sensores en **tres estanques de
acuicultura** de Guntur (India), entre **febrero de 2022 y enero de 2023**, con una lectura cada
**20 minutos**. Incluye siete variables: nitrato, pH, amonio, temperatura, oxígeno disuelto,
turbidez y manganeso. Sus autores lo han usado para clasificar la calidad del agua con modelos de
aprendizaje profundo (Arepalli et al., 2024).

**Problema.** El registro no se puede analizar tal como viene. Proviene de una hoja de cálculo y
presenta columnas numéricas leídas como texto, errores de fórmula, filas vacías, registros sin hora,
fallos de sensor e interrupciones. Antes de estudiar el comportamiento del agua en el tiempo es
necesario convertirlo en un conjunto de datos **confiable, trazable y reproducible**.

## 2. Objetivos

**Objetivo general.** Caracterizar el comportamiento temporal de la calidad del agua en tres
estanques de acuicultura monitoreados por IoT, mediante un flujo reproducible en Python que abarque
desde la depuración del registro hasta su análisis y modelado.

## 3. Alcance y supuestos

**Alcance de este avance (F1–F2):** definición del problema, entorno reproducible, repositorio,
obtención, exploración, limpieza, transformación y validación del archivo `Ponds data.csv`.

**Queda para F3–F4:** análisis de patrones, modelado, evaluación y comunicación de resultados.

**Supuestos principales**
- La cadencia nominal del registro es de 20 minutos (72 lecturas diarias).
- Las marcas de tiempo están en hora local.
- Un valor fuera del rango que un sensor puede medir se trata como fallo de instrumento, no como dato.
- La etiqueta de calidad de la fuente (0/1) se interpreta con 1 como calidad no apta; debe confirmarse antes de modelar.

## 5. Entorno reproducible

El entorno se reconstruye en cualquier equipo con los mismos pasos:

```bash
git clone https://github.com/Macidf/MCDI500-G8-Programaci-n-para-la-Ciencia-de-Datos.git
cd MCDI500-G8-Programaci-n-para-la-Ciencia-de-Datos

python -m venv .venv
.venv\Scripts\activate          # Windows
pip install -r requirements.txt
```

- `requirements.txt` lista las librerías con **versiones exactas**, para que todo el equipo trabaje con el mismo entorno.
- El entorno virtual **no se versiona**: se reconstruye desde `requirements.txt`.
- Los datos originales se descargan desde Kaggle y su procedencia queda registrada en `data/raw/README_DATOS.md`.

## 6. Estructura del repositorio y versionamiento

```
├── README.md              Puerta de entrada: problema, datos, instalación y ejecución
├── requirements.txt       Dependencias con versiones exactas
├── .gitignore             Define qué NO se versiona (entorno, datos, archivos locales)
├── F1/                    Fase 1: definición del proyecto (este notebook)
├── F2/                    Fase 2: obtención, limpieza y validación de los datos
├── src/ponds/             Código fuente reutilizable
├── data/raw/              Datos originales e inmutables
├── data/processed/        Datos limpios, generados por código desde raw
└── docs/                  Decisiones técnicas, guía Git y figuras
```

**Ramas.** `dev` integra el trabajo en curso y `main` recibe la versión estable al cerrar cada fase.

**Commits atómicos.** Un cambio lógico por commit, con mensaje descriptivo y prefijo según su tipo:
`feat` (funcionalidad), `fix` (corrección), `docs` (documentación), `test` (validación) y
`chore` (entorno y configuración). La guía completa está en `docs/guia_git.md`.
